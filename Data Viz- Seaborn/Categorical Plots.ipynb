```python
import seaborn as sns
```

```python
tips = sns.load_dataset('tips')
tips.head()
```

```python
sns.barplot(tips,x='sex',hue='sex',y='total_bill')
```

```python
import numpy as np
sns.barplot(tips,x='sex',hue='sex',y='total_bill',estimator= np.std )
```

```python
sns.countplot(tips,x='sex', hue='sex')
```

```python
sns.boxplot(data=tips,x='day',y='total_bill',hue='smoker')
```

```python
sns.boxplot(data=tips,x='day',y='total_bill',palette='rainbow')
```

```python
sns.boxplot(data=tips,palette='rainbow',orient='h')
```

```python
sns.violinplot(tips,x='day',y='total_bill')
```

```python
sns.violinplot(tips,x='day',y='total_bill', hue='sex',palette='Set1')
```

```python
sns.violinplot(tips,x='day',y='total_bill',split= True, hue='sex',palette='Set1')
```

```python
sns.stripplot(tips,x='day',y='total_bill',jitter=False)
```

```python
sns.stripplot(tips,x='day',y='total_bill',hue='sex',palette='Set1')
```

```python
sns.stripplot(tips,x='day',y='total_bill',hue='sex',palette='Set1',dodge=True)
```

```python
sns.swarmplot(tips,x='day',y='total_bill')
```

```python
sns.swarmplot(tips,x='day',y='total_bill',hue='sex',palette='Set1')
```

```python
sns.swarmplot(tips,x='day',y='total_bill',hue='sex',palette='Set1',dodge=True)
```

```python
sns.violinplot(tips,x='tip',y='day',hue='day',palette='rainbow')
sns.swarmplot(tips,x='tip',y='day',size=3,color='black')
```

```python
sns.catplot(x='sex',y='total_bill',data=tips,kind='bar')
```
