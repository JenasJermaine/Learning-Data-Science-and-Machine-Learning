```python
import seaborn as sns
```

```python
tips=sns.load_dataset('tips')
flights=sns.load_dataset('flights')
```

```python
tips.head()
```

```python
flights.head()
```

```python
tc=tips.corr(numeric_only= True)
```

```python
sns.heatmap(tc)
```

```python
sns.heatmap(tc,cmap='coolwarm',annot=True)
```

```python
flights_pivot= flights.pivot_table(values='passengers',index='month',columns='year',observed= False)
```

```python
flights_pivot.head()
```

```python
sns.heatmap(flights_pivot)
```

```python
sns.heatmap(flights_pivot,cmap='magma',linecolor='white',linewidths=0.5)
```

```python
sns.clustermap(flights_pivot)
```

```python
sns.clustermap(flights_pivot,cmap='coolwarm',standard_scale=1)
```
