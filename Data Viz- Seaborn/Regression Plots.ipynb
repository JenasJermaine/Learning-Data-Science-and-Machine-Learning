```python
import seaborn as sns
```

```python
tips= sns.load_dataset('tips')
```

```python
sns.lmplot(tips,x='total_bill',y='tip')
```

```python
sns.lmplot(tips,x='total_bill',y='tip',hue='sex')
```

```python
sns.lmplot(tips,x='total_bill',y='tip',hue='sex',markers=['o','v'])
```

```python
sns.lmplot(tips,x='total_bill',y='tip',hue='sex',markers=['o','v'],scatter_kws={'s':100})
```

```python
sns.lmplot(tips,x='total_bill',y='tip',col='sex')
```

```python
sns.lmplot(tips,x='total_bill',y='tip',col='sex',row='time')
```

```python
sns.lmplot(tips,x='total_bill',y='tip',col='day',hue='sex',aspect=0.6)
```
