```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.datasets import load_breast_cancer
```

```python
cancer = load_breast_cancer()
```

```python
print(cancer['DESCR'])
```

```python
df= pd.DataFrame(data= cancer['data'], columns = cancer['feature_names'])
```

```python
df.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
scaler.fit(df)
scaled_data = scaler.transform(df)
```

```python
#PCA
from sklearn.decomposition import PCA
```

```python
pca = PCA(n_components= 2)
```

```python
pca.fit(scaled_data)
```

```python
x_pca = pca.transform(scaled_data)
```

```python
scaled_data.shape
```

```python
x_pca.shape
```

```python
plt.figure(figsize = (8,6))
plt.scatter(x_pca[:,0], x_pca[:,1], c= cancer['target'], cmap = 'plasma')
plt.xlabel('First Pricipal Component')
plt.ylabel('Second Principal Component')
```

-To understand which original variables (features) are most represented by each principal component in sklearn.decomposition.PCA, you need to examine the pca.components_ attribute.

**Understand pca.components_:**

-This attribute stores the principal axes in feature space, which are the eigenvectors of the covariance matrix.

-It is a NumPy array with the shape (n_components, n_features).

-Each row represents a principal component, and each column corresponds to an original feature.

-The values within each row are the "loadings" or coefficients, indicating the weight or contribution of each original feature to that specific principal component.

**Interpret the Loadings:**

-Magnitude: The absolute value of a loading indicates the strength of the relationship between an original feature and a principal component. A larger absolute value means a stronger influence.

-Sign: The sign (positive or negative) indicates the direction of the relationship. A positive sign means the feature and component move in the same direction, while a negative sign means they move in opposite directions.

```python
pca.components_
```

```python
df_comp = pd.DataFrame(pca.components_, columns = cancer['feature_names'])
df_comp
```

```python
plt.figure(figsize = (12,6))
sns.heatmap(df_comp, cmap = 'plasma')
```
